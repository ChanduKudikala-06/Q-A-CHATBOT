```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model=SentenceTransformer('all-MiniLM-L6-v2')

text="""Langchain is a framework for building applications with LLM.
Langchain provides modular abstraction to combine llms
You can create chains,agents,memory and retriever"""
```

```python
#Step1 Split into sentences

from torch import embedding


sentences=[s.strip()for s in text.split('\n') if s.strip()]

embedding=model.encode(sentences)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain.chat_models import init_chat_model
from langchain_core.runnables import RunnableLambda,RunnableMap
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
#Semantic chunker with Langchain

from langchain_experimental.text_splitter import SemanticChunker
```

```python
#Load the documents

from langchain_community.document_loaders import TextLoader


loader=TextLoader("text.txt")

docs=loader.load()

#Intializing the embedding model

embedding=OpenAIEmbeddings()

#Create the semantic chunker
chunker=SemanticChunker(embedding)

chunks=chunker.split_documents(docs)

for i,chunk in enumerate(chunks):
    print(f"\n chunk {i+1}:\n {chunk.page_content}")
```
