```python
#Hybrid Retriever-Combining Dense and Sparse Retrieval

from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_community.retrievers import BM25Retriever
#from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
```

```python
new_doc1=Document(
    page_content="new_document1",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
new_doc2=Document(
    page_content="Langchain is a framework for building applications with LLM.",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
new_doc3=Document(
    page_content="new_document3",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
new_doc4=Document(
    page_content="new_document1",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
new_doc5=Document(
    page_content="Langchain provides modular abstraction to combine llms",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
new_doc6=Document(
    page_content="You can create chains,agents,memory and retriever",
    metadata={"source":"manual_addition","topic":"reinforcement_learning"}
)
documents=[new_doc1,new_doc2,new_doc3,new_doc4,new_doc5,new_doc6]
```

```python
import os

from dotenv import  load_dotenv

load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
embedding=OpenAIEmbeddings()

#DenseRetriever
dense_vector_store=FAISS.from_documents(documents,embedding)
dense_retriever=dense_vector_store.as_retriever()
```

```python
#Sparse Retriever(BM25)

sparse_retriever=BM25Retriever.from_documents(documents)
sparse_retriever.k=1 #top 3 documents to retrievve


#Combining Dense and Sparse Retrieval

def ensemble_retriever(query):

    dense_docs=dense_retriever.invoke(query)
    sparse_docs=dense_retriever.invoke(query)
    hybrid_docs=dense_docs + sparse_docs


    return hybrid_docs


```

```python
query="What is langchain"
a=ensemble_retriever(query)
```

```python
#Re Ranking Strategies
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.chat_models import init_chat_model
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
#load text file
loader=TextLoader("text.txt")

raw_docs=loader.load()
splits=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)

doc_splits=splits.split_documents(raw_docs)
```

```python
#user query

user_query="InfoWorld described LangChain as a software development kit"

from langchain_community.vectorstores import FAISS

from langchain_core import retrievers
from langchain_openai import OpenAIEmbeddings
embedding_model=OpenAIEmbeddings()

vector_Store=FAISS.from_documents(doc_splits,embedding_model)
retriever=vector_Store.as_retriever(search_kwargs={"k":8})
retriever
```

```python
#Prompt and use llm
import os

from dotenv import  load_dotenv
from langchain.chat_models import init_chat_model
from langchain_openai import ChatOpenAI

load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")


os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
llm=ChatOpenAI(model="gpt-4o")
```

```python
prompt=PromptTemplate.from_template(
    """You are a helpful assistant.Your task is to rank the follwing documents from most to relavant

    user_query:"{user_query}"

    Documents:{documents}

    Instructions:
    -Rank about the relevance of the each document to the user's query
    Return a list of document indices in ranked order,starting from the most relvant.
    Output format :comma separated document indices (eg,2,1,3,0...)

    """
)
```

```python
retriever_docs=retriever.invoke(user_query)
retriever_docs
```

```python
doc_lines=[f"{index+1}. {doc.page_content}" for index, doc in enumerate(retriever_docs)]
formated_docs="\n".join(doc_lines)

formated_docs
```

```python
from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()

chain=prompt|llm|parser

response=chain.invoke(
    {
        "user_query":user_query,
        "documents":formated_docs
        
    }
)

```
